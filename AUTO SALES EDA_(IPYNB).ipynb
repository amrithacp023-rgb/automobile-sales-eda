```python
df['ORDERDATE'] = pd.to_datetime(
    df['ORDERDATE'],
    dayfirst=True,
    errors='coerce'
)
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['SALES'], bins=30, kde=True)

plt.title('Distribution of Automobile Sales')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

```python
product_sales = df.groupby('PRODUCTLINE')['SALES'].sum()

product_sales
```

```python
plt.figure(figsize=(10,5))

sns.barplot(x=product_sales.index,
            y=product_sales.values)

plt.title('Total Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales')

plt.xticks(rotation=45)
plt.show()
```

```python
num_df = df.select_dtypes(include='number')

num_df.columns
```

```python
corr = df[['QUANTITYORDERED',
           'PRICEEACH',
           'SALES',
           'MSRP']].corr()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title('Correlation Between Automobile Sales Variables')

plt.show()
```

```python
df['YEAR'] = df['ORDERDATE'].dt.year
```

```python
yearly_sales = df.groupby('YEAR')['SALES'].sum()

yearly_sales
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x=yearly_sales.index,
            y=yearly_sales.values)

plt.title('Yearly Automobile Sales')
plt.xlabel('Year')
plt.ylabel('Total Sales')

plt.show()
```

```python
country_sales = df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)

country_sales
```

```python
plt.figure(figsize=(12,6))

sns.barplot(x=country_sales.index,
            y=country_sales.values)

plt.title('Total Sales by Country')
plt.xlabel('Country')
plt.ylabel('Total Sales')

plt.xticks(rotation=60)
plt.show()
```

```python
deal_sales = df.groupby('DEALSIZE')['SALES'].sum()

deal_sales
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x=deal_sales.index,
            y=deal_sales.values)

plt.title('Sales by Deal Size')
plt.xlabel('Deal Size')
plt.ylabel('Total Sales')

plt.show()
```

```python
df['MONTH'] = df['ORDERDATE'].dt.month_name()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_sales = df.groupby('MONTH')['SALES'].sum()
monthly_sales = monthly_sales.reindex(month_order)

plt.figure(figsize=(12,5))
sns.barplot(x=monthly_sales.index, y=monthly_sales.values)

plt.title('Monthly Automobile Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
status_counts = df['STATUS'].value_counts()

plt.figure(figsize=(10,5))
sns.barplot(x=status_counts.index, y=status_counts.values)

plt.title('Order Status Distribution')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.show()
```

## 1. Project overview

The main objective of this EDA was to understand the automobile company's sales performance using different charts. We analysed sales across years, months, countries, product lines, deal sizes, and order statuses.

## 2. Key insights from the analysis

1\. Yearly Sales: 2019 recorded the highest sales, while 2020 showed a significant decline compared with previous years.

2\. Product Line: We analysed sales across different product categories to understand which product lines contribute most to total revenue.

3\. Country-wise Sales: USA was identified as a major sales market. Ireland and France could be investigated further for potential business opportunities.

4\. Deal Size: Medium deals contributed significantly more sales than large deals. Large deals need further investigation to understand their lower contribution.

5\. Monthly Sales: November recorded the highest sales, while June recorded the lowest. Sales were relatively consistent across other months, with a noticeable jump in November.

6\. Order Status: Shipped orders had the highest count, while disputed orders had the lowest count. This indicates that most orders had reached the shipping stage.

## 3. Important key points

- 2019 was the strongest sales year in the dataset.
- 2020 needs further investigation because of its lower sales and fewer records.
- USA is an important geographical market.
- Medium-sized deals are a major contributor to revenue.
- November shows a noticeable sales opportunity.
- Disputed orders should be monitored to reduce customer-related issues.
- Large deals need more attention to understand their performance.

## 4. Final conclusion

The EDA helped us understand the company's overall sales performance and identify important business patterns. The analysis highlighted strong performance in 2019, the importance of the USA market, the contribution of medium-sized deals, and the November sales peak.

The company should focus on maintaining strong markets, investigating underperforming areas, and improving its sales strategy using these findings.

In simple words: EDA helped us understand what is happening in the company's sales, where the company is performing well, and which areas need more attention.
